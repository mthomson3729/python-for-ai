```python
import os
import pandas as pd
from dotenv import load_dotenv
import cbbd
from cbbd.rest import ApiException
load_dotenv()

#def get_data():
api_key = os.getenv('CBBD_API_KEY')
if not api_key:
    raise ValueError("CBBD_API_KEY not found. Ensure it is defined in your .env file.")

configuration = cbbd.Configuration(
        access_token=api_key,
    )

try:
    with cbbd.ApiClient(configuration) as api_client:
        games_api = cbbd.GamesApi(api_client)
        games = games_api.get_games(season=2024, team='Duke')

        if games:
            games_dict = [game.to_dict() for game in games]
            df = pd.DataFrame(games_dict)

            print(df.head)
except ApiException as error:
    print(f'College Basketball Data API request failed: {error}')

df.columns.tolist()

#def to_csv():
    
```

The main philosophy behind predicting any of these games wil revolve around *TWO* main ideas. 1) The True Offesnive, Defensive, and March-Preparedness of a Team 2) The Value at Risk of the Team playing. 

Philosohpy Point 1: Like any good team, you need to be simply good at Basketball. Can you score more points than your opponet? Can you stop a good team from scoring? These are covered in a very analytical sense, using stats such as the Adj Offensive Ranking, Adj Deffensive Ranking, etc. The key idea behind using the Adj side of things comes down to the idea that SOS is a very real thing in college basketball. A mid SEC team would kill the best team in the MACC, so it's important to check that the opponets a team are playing and adjust the numbers accordingly. This is less of a psychological startegy but one that simply gets the statistically best teams priotitzed. 

Philosophy Point 2: This is much less of a analytical strategy, but one that covers the shifting college landscape and not so beloved second round of the NBA draft.
